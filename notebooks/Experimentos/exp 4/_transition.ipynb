```python
import numpy as np
```

```python
action = 0

masspole = 0.1
length = 0.5  # actually half the pole's length

gravity = 9.8
masscart = 1.0
force_mag = 10.0
tau = 0.02  # seconds between state updates
```

```python
def transition(state, action):
    x, x_dot, theta, theta_dot = state
    action_dir = (-1,1)[action] 


    temp = ((force_mag*action_dir) + (masspole * length) * np.square(theta_dot) * np.sin(theta)) / (masspole + masscart)
    print(temp)

    thetaacc = (
        (gravity * np.sin(theta) - np.cos(theta) * temp) / 
        (length * (4.0 / 3.0 - masspole * np.square(np.cos(theta)) / (masspole + masscart)))
    )
    xacc = temp - (masspole * length) * thetaacc * np.cos(theta) / (masspole + masscart)

    x = x + tau * x_dot
    x_dot = x_dot + tau * xacc
    theta = theta + tau * theta_dot
    theta_dot = theta_dot + tau * thetaacc
    print(thetaacc)
    return np.array((x, x_dot, theta, theta_dot), dtype=np.float64)
```

```python
def get_pole_length(state_1, action, state_2):
    x_1, x_dot_1, theta_1, theta_dot_1 = state_1
    x_2, x_dot_2, theta_2, theta_dot_2 = state_2
    action_dir = (-1,1)[action] 

    xacc = (x_dot_2 - x_dot_1)/tau
    theta_dot = (theta_2 - theta_1)/tau
    thetaacc = (theta_dot_2 - theta_dot_1)/tau
    return (3*gravity*np.sin(theta_1)*(masspole+masscart) - 3*force_mag*action_dir*np.cos(theta_1))/(thetaacc*(4-masspole*np.square(np.cos(theta_1))) + 3*masspole*np.square(theta_dot_1)*np.sin(theta_1)*np.cos(theta_1))
```

```python
def get_pole_mass(state_1, action, state_2):
    x_1, x_dot_1, theta_1, theta_dot_1 = state_1
    x_2, x_dot_2, theta_2, theta_dot_2 = state_2
    action_dir = (-1,1)[action] 

    xacc = (x_dot_2 - x_dot_1)/tau
    theta_dot = (theta_2 - theta_1)/tau
    thetaacc = (theta_dot_2 - theta_dot_1)/tau
    return (3*(masscart*gravity*np.sin(theta_1)- force_mag*action_dir*np.cos(theta_1)) - 4*length*thetaacc)/(length*np.square(np.cos(theta_1)) + 3*np.sin(theta_1)*(length*np.square(theta_dot_1)*np.cos(theta_1)-gravity))
```

```python
masspole, length = 0.1, 1.6

s = (-0.082, -0.483, -0.166, -0.299)
a = 0

s_ = transition(s, a) #(-0.091, -0.668, -0.172, -0.225)
s_
```

```python
get_pole_length(s,a,s_)
```

```python
get_pole_mass(s,a,s_)
```
