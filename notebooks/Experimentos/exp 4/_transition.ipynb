$ s = (x, \dot{x}, \theta, \dot{\theta}) $

$
\theta acc = \frac{g \times sin(\theta) - cos(\theta) \times \frac{fm \times a + mp \times l * \dot{\theta}^2 * sin(\theta)}
{mp + mc}}
{l \times \frac{\frac{4}{3} - mp \times cos(\theta)^2}
{mp + mc}}
$

$
xacc = \frac{fm \times a + mp \times l * \dot{\theta}^2 * sin(\theta)}
{mp + mc} - \frac{mp \times l \times \theta acc \times cos(\theta)}{mp + mc}
$

$x = x + \tau \times \dot{x}$

$\dot{x} = \dot{x} + \tau \times xacc$

$\theta = x + \tau \times \dot{\theta}$

$\dot{\theta} = \dot{\theta} + \tau \times \theta acc$

$mp$ - Pole's mass <br>
$l$ - Pole's lenght <br>
$mc$ - Cart's mass <br>
$g$ - Gravity <br>
$\tau$ - Change in time (Seconds between states) <br>
$fm$ - Force's magnitude <br>
$a$ - Force's direction, -1 or 1 (action) <br>

```python
import numpy as np
```

```python
action = 0

masspole = 0.1
length = 0.5  # actually half the pole's length

gravity = 9.8
masscart = 1.0
force_mag = 10.0
tau = 0.02  # seconds between state updates
```

```python
def transition(state, action):
    x, x_dot, theta, theta_dot = state
    action_dir = (-1,1)[action] 

    thetaacc = (
        (
            gravity * np.sin(theta) - np.cos(theta) * (
                ((force_mag*action_dir) + (masspole * length) * np.square(theta_dot) * np.sin(theta)) / 
                (masspole + masscart)
            )
        ) / 
        (length * (4.0 / 3.0 - masspole * np.square(np.cos(theta)) / (masspole + masscart)))
    )
    xacc = (
        ((force_mag*action_dir) + (masspole * length) * np.square(theta_dot) * np.sin(theta)) / 
        (masspole + masscart)
    ) - (masspole * length) * thetaacc * np.cos(theta) / (masspole + masscart)

    x = x + tau * x_dot
    x_dot = x_dot + tau * xacc
    theta = theta + tau * theta_dot
    theta_dot = theta_dot + tau * thetaacc

    return np.array((x, x_dot, theta, theta_dot), dtype=np.float64)
```

```python
masspole, length = 0.1, 1.6
transition((-0.082, -0.483, -0.166, -0.299), 0) #(-0.091, -0.668, -0.172, -0.225)
```
