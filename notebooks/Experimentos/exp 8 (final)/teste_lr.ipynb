### Least Squares

$\hat{w} = (M^T M)^{-1} M^T Y$

```python
import numpy as np

m = np.array([
    [-0.011, -0.058, -0.013, -0.015, -0.013], 
    [-0.584, -0.587, -0.585, -0.547, -0.571],
    [0.025,  0.019,  0.024,  0.024,  0.025],
    [0.932,  1.158,  0.796,  0.943,  0.651],
])
y = np.array([-0.013, -0.573, 0.024, 0.704])

w = np.array([0.,0.,.1,0.,.9])
print('err:', np.pow(np.sum(w*m, axis=1) - y, 2))
print('value:', np.sum(w*m, axis=1))

# w, residuals, rank, s_values = np.linalg.lstsq(A, b, rcond=None)
w_, residuals, rank, s_values = np.linalg.lstsq(m, y)
print('est: ', w_)
print('err:', np.pow(np.sum(w_*m, axis=1) - y, 2))
print('value:', np.sum(w_*m, axis=1))
```
