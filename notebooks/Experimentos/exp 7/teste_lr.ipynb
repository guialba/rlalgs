### Least Squares

$\hat{w} = (M^T M)^{-1} M^T Y$

```python
m = np.array([[3., 2., 1.], [6., 5., 4.]])
y = np.array([2., 5.])

w = np.array([0.,1.,0.])
print('err:', np.pow(np.sum(w*m, axis=1) - y, 2))
print('value:', np.sum(w*m, axis=1))

# w, residuals, rank, s_values = np.linalg.lstsq(A, b, rcond=None)
w_, residuals, rank, s_values = np.linalg.lstsq(m, y)
print('est: ', w_)
print('err:', np.pow(np.sum(w_*m, axis=1) - y, 2))
print('value:', np.sum(w_*m, axis=1))
```
