```python
from pmbrl.data import Experiment_Data
import numpy as np
import matplotlib.pyplot as plt
```

# Mass Pole

```python
n=100
shape, scale, factor = 2., 12., 100
print("mean", (shape*scale)/factor) 
s = np.random.gamma(shape, scale, n)/factor

fig, axs = plt.subplots(1, 1, figsize=(18,4))
axs.hist(s, bins=10)
plt.show()
```

# Stats

```python
train.raw_data.head()
```

```python
from pmbrl.data import get_data_expanded

expansions = {
    'p': ['p0', 'p1'],
    's': ['s0', 's1', 's2', 's3'],
    's_': ['s_0', 's_1', 's_2', 's_3'],
}

df_train = get_data_expanded(train.raw_data, expansions)
df_test = get_data_expanded(test.raw_data, expansions)
df_train.head()
```

# Params and State stats

## Train

```python
df_train_agg = df_train[['s0', 's1', 's2', 's3', 'p0', 'p1']].agg(['nunique', 'mean', 'std', 'min', 'max'])
df_train_agg
```

```python
fig, axs = plt.subplots(1, 2, figsize=(18,4))
axs[0].hist(df_train['p0'], bins=10)
axs[1].hist(df_train['p1'], bins=10)

axs[0].set_title('mp')
axs[1].set_title('l')
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(18,4))
axs[0].hist(df_test['p0'], bins=10)
axs[1].hist(df_test['p1'], bins=10)

axs[0].set_title('mp')
axs[1].set_title('l')
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(18,4))
axs[0][0].hist(df_train['s0'], bins=50)
axs[0][1].hist(df_train['s1'], bins=50)
axs[1][0].hist(df_train['s2'], bins=50)
axs[1][1].hist(df_train['s3'], bins=50)

axs[0][0].set_title("s0")
axs[0][1].set_title("s1")
axs[1][0].set_title("s2")
axs[1][1].set_title("s3")
plt.show()
```

### First STep

```python
fig, axs = plt.subplots(2, 2, figsize=(18,4))
axs[0][0].hist(df_train.loc[df_train['step']==0]['s0'], bins=50)
axs[0][1].hist(df_train.loc[df_train['step']==0]['s1'], bins=50)
axs[1][0].hist(df_train.loc[df_train['step']==0]['s2'], bins=50)
axs[1][1].hist(df_train.loc[df_train['step']==0]['s3'], bins=50)

axs[0][0].set_title("s0")
axs[0][1].set_title("s1")
axs[1][0].set_title("s2")
axs[1][1].set_title("s3")
plt.show()
```

### Last Step

```python
fig, axs = plt.subplots(2, 2, figsize=(18,4))

get_max = lambda i: df_train.loc[df_train['episode']==i].step.max()
get_last_values = lambda i: df_train.loc[(df_train['episode']==i) & (df_train['step']==get_max(i))][['s0','s1', 's2','s3', 'p0', 'p1']].values[0]
maxes = df_train.groupby("episode").step.max().reset_index()
maxes[['s0','s1', 's2','s3', 'p0', 'p1']] = maxes.apply(lambda row: get_last_values(row['episode']), axis=1, result_type='expand')

axs[0][0].hist(maxes['s0'], bins=50)
axs[0][1].hist(maxes['s1'], bins=50)
axs[1][0].hist(maxes['s2'], bins=50)
axs[1][1].hist(maxes['s3'], bins=50)

axs[0][0].set_title("s0")
axs[0][1].set_title("s1")
axs[1][0].set_title("s2")
axs[1][1].set_title("s3")
plt.show()
```
